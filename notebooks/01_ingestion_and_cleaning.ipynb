#### **Environment Setup & Imports**

```python
import os
import sys
import pandas as pd
from pathlib import Path

# Ensure root directory is in sys.path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

from src.loader import MatchLensLoader

# Initialize Loader
loader = MatchLensLoader(raw_data_dir="../data/raw")
print("MatchLensLoader initialized successfully.")
```

#### **Batch Download Raw Match Payloads**

```python
# Fetch and display available competitions
df_competitions = loader.download_competitions()
print(f"Total competitions available: {len(df_competitions)}")

df_competitions[['competition_id', 'season_id', 'competition_name', 'season_name']].head(10)
```

#### **Download Match Schedules**

```python
COMP_ID = 16
SEASON_ID = 27

df_matches = loader.download_matches(competition_id=COMP_ID, season_id=SEASON_ID)
print(f"Total matches retrieved for competition {COMP_ID}, season {SEASON_ID}: {len(df_matches)}")

# Display a sample of matches
df_matches[['match_id', 'match_date', 'home_team', 'away_team', 'home_score', 'away_score']].head()
```

#### **Batch Download Raw Match Payloads - Events & Lineups**

```python
# Extract match IDs from the downloaded schedule dataframe
match_ids = df_matches['match_id'].tolist()
print(f"Total matches available in schedule: {len(match_ids)}")

target_match_ids = match_ids[:10] 

print(f"Starting batch download of raw events and lineups for {len(target_match_ids)} matches...")
for match_id in target_match_ids:
    loader.download_match_payloads(match_id)

print("Batch ingestion and raw caching complete! Check your data/raw/events/ and data/raw/lineups/ folders.")
```
